```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

create a simple graph


```python
x=np.array([1,8])
y=np.array([3,7])
plt.plot(x,y)
plt.show
```

create of labels  on the graph

```python
x=[1,3,5,6,8,10]
y=[4,6,3,7,5,5]
plt.figure(figsize=(8,5))
plt.plot(x,y,color='blue',linestyle=':',marker='s',linewidth=3)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("line chart")
plt.legend(['my data sets'])
plt.grid(True)
plt.show()
```

bar graph


```python
categories=['A','B','C','D','E']
values=[5,6,3,4,10]
plt.figure(figsize=(8,5))
plt.bar(categories,values,color='green')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Bar chart')
plt.show()
```

histogram graph

```python
data=[23,5,9,68,76,65,2,3,5,67,7,6,12,8,5,23,8,8,55,46,7,75,4,35,67,77,7,6,4,67,4,98,5,6,7,6,54,5,6,56,7,34,65,6,7,7,85,6,9]
plt.hist(data, bins=16, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title(' Histogram Example')
plt.show()
```

scatter graph

```python
pt=[4,56,7,78,8,5,8,9,9,9,6,5]
yh=[3,8,6,5,4,8,7,67,7,77,8,5]
plt.scatter(pt,yh)
plt.show()
```

pie cart

```python
labels=['a','s','k','d']
sizes=[2,6,5,7]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Simple Pie Chart Example')
plt.axis('equal')
```

box plot



```python
data=[7,7,5,4,7,4,3,7,98,3,5,54,4,7]
plt.boxplot(data)
plt.title('Simple Box Plot Example')
plt.ylabel('Values')
plt.show()


```

heat map

```python

data = np.random.rand(5, 10)


plt.imshow(data, cmap='viridis', interpolation='nearest')
plt.colorbar()
plt.title('Simple Heatmap Example')
plt.show()
```
